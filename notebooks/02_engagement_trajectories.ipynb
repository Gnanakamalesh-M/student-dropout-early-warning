# 02 — Engagement: does disengagement precede withdrawal?

This notebook tests the premise the whole system rests on. All aggregates
respect the as-of boundary from `docs/TASK_SPEC.md`.

```python
import pandas as pd

pd.set_option("display.width", 200)

from dropout_ews.data.checkpoints import build_checkpoint_rows
from dropout_ews.eda import analysis as eda

population, report = build_checkpoint_rows()
print(f"{len(population):,} rows, {population['label'].mean():.2%} positive")

windows = eda.engagement_windows(population)
```

## Engagement aligned to the withdrawal event

Aligning on the event rather than calendar time is what reveals whether
disengagement *precedes* withdrawal. Finding: it does, but gradually, with
most of the decline inside the final two weeks.

```python
aligned = eda.aligned_engagement_before_event(population, weeks_before=12)
aligned
```

```python
ax = aligned["median_clicks"].plot(marker="o", figsize=(7, 4))
ax.invert_xaxis()
ax.set_xlabel("Weeks before withdrawal")
ax.set_ylabel("Median weekly clicks")
ax.grid(alpha=0.3)
```
