# 03 — Feature screening, baselines, and subgroup base rates

Univariate screening to inform the Phase 4 allowlist, the trivial-rule bar any
model must beat, and subgroup base rates recorded *before* modelling so any
later disparity is interpretable.

```python
import pandas as pd

pd.set_option("display.width", 200)

from dropout_ews.data.checkpoints import build_checkpoint_rows
from dropout_ews.eda import analysis as eda

population, report = build_checkpoint_rows()
print(f"{len(population):,} rows, {population['label'].mean():.2%} positive")

windows = eda.engagement_windows(population)
```

## Univariate discriminative power

AUC below 0.5 is the expected direction (more activity, less risk); ranking
uses distance from 0.5. Note that the crude trend features rank **last** —
see Finding 2 in `docs/EDA_FINDINGS.md`.

```python
eda.discriminative_power(windows)
```

## Stability across checkpoints

```python
for feature in ("clicks_7d", "days_since_last_activity", "clicks_delta_30d"):
    print(feature)
    print(eda.discriminative_power_by_checkpoint(windows, feature).to_string(index=False))
    print()
```

## Trivial rule baselines

The bar. Best rule reaches ~5% precision against a 3.02% base rate while
flagging 22% of the cohort — operationally unusable, which is why a model has
real work to do.

```python
eda.trivial_rule_baselines(windows)
```

## Subgroup base rates

These are **base rates, not model errors**. A calibrated model will flag
higher-base-rate groups more often, which is not by itself unfairness. These
attributes are excluded from the feature matrix (`docs/ETHICS.md`).

```python
eda.subgroup_base_rates(windows)
```
